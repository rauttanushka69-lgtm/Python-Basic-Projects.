```python
import pandas as pd

# Load datasets
heart = pd.read_csv("heart.csv")
housing = pd.read_csv("Housing.csv")

print("Heart Dataset")
display(heart.head())

print("Housing Dataset")
display(housing.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

# ------------------ HEART DATASET ------------------

heart = heart.dropna()

# Convert categorical columns to numbers
le = LabelEncoder()

for col in heart.select_dtypes(include='object').columns:
    heart[col] = le.fit_transform(heart[col])

X_cls = heart.drop("HeartDisease", axis=1)
y_cls = heart["HeartDisease"]

scaler_cls = StandardScaler()
X_cls = scaler_cls.fit_transform(X_cls)

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls, y_cls, test_size=0.2, random_state=42
)

# ------------------ HOUSING DATASET ------------------

housing = housing.dropna()

for col in housing.select_dtypes(include='object').columns:
    housing[col] = le.fit_transform(housing[col])

X_reg = housing.drop("price", axis=1)
y_reg = housing["price"]

scaler_reg = StandardScaler()
X_reg = scaler_reg.fit_transform(X_reg)

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

print("Preprocessing Completed Successfully")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Logistic Regression Model
lr = LogisticRegression()
lr.fit(X_train_cls, y_train_cls)

# Prediction
y_pred = lr.predict(X_test_cls)

# Results
print("Accuracy:", accuracy_score(y_test_cls, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_cls, y_pred))

print("\nClassification Report:")
print(classification_report(y_test_cls, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Decision Tree Model
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train_cls, y_train_cls)

# Prediction
y_pred = dt.predict(X_test_cls)

# Results
print("Accuracy:", accuracy_score(y_test_cls, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_cls, y_pred))

print("\nClassification Report:")
print(classification_report(y_test_cls, y_pred))
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# SVM Model
svm = SVC()
svm.fit(X_train_cls, y_train_cls)

# Prediction
y_pred = svm.predict(X_test_cls)

# Results
print("Accuracy:", accuracy_score(y_test_cls, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_cls, y_pred))

print("\nClassification Report:")
print(classification_report(y_test_cls, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# KNN Model
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_cls, y_train_cls)

# Prediction
y_pred = knn.predict(X_test_cls)

# Results
print("Accuracy:", accuracy_score(y_test_cls, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_cls, y_pred))

print("\nClassification Report:")
print(classification_report(y_test_cls, y_pred))
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Naive Bayes Model
nb = GaussianNB()
nb.fit(X_train_cls, y_train_cls)

# Prediction
y_pred = nb.predict(X_test_cls)

# Results
print("Accuracy:", accuracy_score(y_test_cls, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_cls, y_pred))

print("\nClassification Report:")
print(classification_report(y_test_cls, y_pred))
```

```python
%%writefile app.py

import streamlit as st
import joblib
import numpy as np

st.set_page_config(page_title="Session 24 AIML", layout="centered")

st.title("Session 24 - Multi Model Prediction App")

problem = st.selectbox(
    "Select Problem Type",
    ["Classification", "Regression"]
)

if problem == "Classification":

    model = joblib.load("best_classification_model.pkl")
    scaler = joblib.load("classification_scaler.pkl")

    st.header("Heart Disease Prediction")

    age = st.number_input("Age", 1, 100, 40)
    sex = st.selectbox("Sex", [0, 1])
    chest = st.selectbox("Chest Pain Type", [0, 1, 2, 3])
    bp = st.number_input("Resting BP", 80, 250, 120)
    chol = st.number_input("Cholesterol", 100, 600, 200)
    sugar = st.selectbox("Fasting Blood Sugar", [0, 1])
    ecg = st.selectbox("Rest ECG", [0, 1, 2])
    hr = st.number_input("Max Heart Rate", 60, 220, 150)
    angina = st.selectbox("Exercise Angina", [0, 1])
    oldpeak = st.number_input("Oldpeak", 0.0, 10.0, 1.0)
    slope = st.selectbox("ST Slope", [0, 1, 2])

    if st.button("Predict"):
        data = np.array([[age, sex, chest, bp, chol,
                          sugar, ecg, hr, angina,
                          oldpeak, slope]])

        data = scaler.transform(data)
        pred = model.predict(data)

        if pred[0] == 1:
            st.error("Heart Disease Detected")
        else:
            st.success("No Heart Disease")
```
