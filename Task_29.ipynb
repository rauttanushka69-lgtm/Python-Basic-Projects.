```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load dataset
df = pd.read_csv(
    "train.txt",
    sep=";",
    names=["text", "emotion"]
)

# Separate features and target
X = df["text"]
y = df["emotion"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Print shapes
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_train.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create CountVectorizer
cv = CountVectorizer()

# Fit and transform training data
X_train_bow = cv.fit_transform(X_train)

# Transform test data
X_test_bow = cv.transform(X_test)

# Print shapes
print("X_train_bow shape:", X_train_bow.shape)
print("X_test_bow shape:", X_test_bow.shape)

# Display first 20 feature names
print("First 20 feature names:")
print(cv.get_feature_names_out()[:20])
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Create and train the model
model = MultinomialNB()
model.fit(X_train_bow, y_train)

# Make predictions
y_pred = model.predict(X_test_bow)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Print accuracy
print("Bag of Words + MultinomialNB Accuracy:", accuracy)
```

```python
# Print total vocabulary size
print("Total Vocabulary Size:", len(cv.vocabulary_))

# Display any 15 words
print("\n15 Vocabulary Words:")
print(list(cv.vocabulary_.keys())[:15])

# Convert one sample document into BoW vector
sample_text = X_train.iloc[0]
sample_vector = cv.transform([sample_text])

print("\nSample Document:")
print(sample_text)

print("\nBag of Words Vector:")
print(sample_vector.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create CountVectorizer with unigrams + bigrams
cv_bigram = CountVectorizer(ngram_range=(1, 2))

# Fit and transform training data
X_train_bigram = cv_bigram.fit_transform(X_train)

# Transform test data
X_test_bigram = cv_bigram.transform(X_test)

# Print shapes
print("X_train_bigram shape:", X_train_bigram.shape)
print("X_test_bigram shape:", X_test_bigram.shape)

# Display first 20 features (includes bigrams)
print("\nSome Bigram Features:")
print(cv_bigram.get_feature_names_out()[:20])
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Train model using bigram features
bigram_model = MultinomialNB()
bigram_model.fit(X_train_bigram, y_train)

# Make predictions
y_pred_bigram = bigram_model.predict(X_test_bigram)

# Calculate accuracy
bigram_accuracy = accuracy_score(y_test, y_pred_bigram)

# Print accuracy
print("Bigram Accuracy:", bigram_accuracy)

# Compare with unigram accuracy from Q3
print("Unigram Accuracy:", accuracy)
print("Bigram Accuracy:", bigram_accuracy)
```

```python
from sklearn.feature_extraction.text import TfidfVecto rizer

# Create TF-IDF Vectorizer
tfidf = TfidfVectorizer()

# Fit and transform training data
X_train_tfidf = tfidf.fit_transform(X_train)

# Transform test data
X_test_tfidf = tfidf.transform(X_test)

# Print shapes
print("X_train_tfidf shape:", X_train_tfidf.shape)
print("X_test_tfidf shape:", X_test_tfidf.shape)

# Display first 15 feature names
print("\nFirst 15 TF-IDF Feature Names:")
print(tfidf.get_feature_names_out()[:15])
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Create and train model
tfidf_model = MultinomialNB()
tfidf_model.fit(X_train_tfidf, y_train)

# Make predictions
y_pred_tfidf = tfidf_model.predict(X_test_tfidf)

# Calculate accuracy
tfidf_accuracy = accuracy_score(y_test, y_pred_tfidf)

# Print accuracy
print("TF-IDF + MultinomialNB Accuracy:", tfidf_accuracy)
```

```python
import pandas as pd

# Create comparison table
comparison = pd.DataFrame({
    "Vectorizer": [
        "Bag of Words (Unigrams)",
        "Bag of Words (Unigrams + Bigrams)",
        "TF-IDF"
    ],
    "Accuracy": [
        accuracy,
        bigram_accuracy,
        tfidf_accuracy
    ]
})

print(comparison)

# Observation
best_method = comparison.loc[comparison["Accuracy"].idxmax()]

print("\nObservation:")
print(f"{best_method['Vectorizer']} performed the best with an accuracy of {best_method['Accuracy']:.4f}.")
```

```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Load dataset
df = pd.read_csv("train.txt", sep=";", names=["text", "emotion"])

# Features and target
X = df["text"]
y = df["emotion"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# CountVectorizer
cv = CountVectorizer()
X_train_bow = cv.fit_transform(X_train)
X_test_bow = cv.transform(X_test)

bow_model = MultinomialNB()
bow_model.fit(X_train_bow, y_train)
bow_pred = bow_model.predict(X_test_bow)
bow_accuracy = accuracy_score(y_test, bow_pred)

# TF-IDF
tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

tfidf_model = MultinomialNB()
tfidf_model.fit(X_train_tfidf, y_train)
tfidf_pred = tfidf_model.predict(X_test_tfidf)
tfidf_accuracy = accuracy_score(y_test, tfidf_pred)

# Compare
print("Bag of Words Accuracy:", bow_accuracy)
print("TF-IDF Accuracy:", tfidf_accuracy)

# Save best model and vectorizer
if bow_accuracy >= tfidf_accuracy:
    joblib.dump(bow_model, "best_model.pkl")
    joblib.dump(cv, "best_vectorizer.pkl")
    print("Best Model: Bag of Words")
else:
    joblib.dump(tfidf_model, "best_model.pkl")
    joblib.dump(tfidf, "best_vectorizer.pkl")
    print("Best Model: TF-IDF")

print("Best model and vectorizer saved successfully.")
```
