```python
import pandas as pd

# Load the dataset
df = pd.read_csv(
    "train.txt",
    sep=";",
    names=["text", "emotions"]
)

# Display the first 10 rows
print("First 10 Rows:")
print(df.head(10))

# Display dataset shape
print("\nDataset Shape:")
print(df.shape)

# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder object
le = LabelEncoder()

# Find and print unique emotion labels
print("Unique Emotion Labels:")
print(df["emotions"].unique())

# Encode emotion labels
df["emotion_encoded"] = le.fit_transform(df["emotions"])

# Create mapping dictionary
label_mapping = dict(zip(le.classes_, le.transform(le.classes_)))

print("\nEmotion Label Mapping:")
print(label_mapping)

# Display first 10 rows with encoded labels
print("\nDataset with Encoded Labels:")
print(df[["text", "emotions", "emotion_encoded"]].head(10))
```

```python
# Show 5 sample texts before lowercasing
print("Before Lowercasing:\n")
print(df["text"].head(5))

# Convert text to lowercase
df["text"] = df["text"].str.lower()

# Show 5 sample texts after lowercasing
print("\nAfter Lowercasing:\n")
print(df["text"].head(5))

# Short note
print("\nWhy is lowercasing important in NLP?")
print("Lowercasing converts all text into a common format. It treats words like 'Happy' and 'happy' as the same word, reduces vocabulary size, and improves text processing and model accuracy.")
```

```python
import string

# Function to remove punctuation
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

# Show 5 samples before removing punctuation
print("Before Removing Punctuation:\n")
print(df["text"].head(5))

# Apply the function
df["text"] = df["text"].apply(remove_punctuation)

# Show 5 samples after removing punctuation
print("\nAfter Removing Punctuation:\n")
print(df["text"].head(5))
```

```python
import re

# Function to remove numbers
def remove_numbers(text):
    return re.sub(r'\d+', '', text)

# Show 5 samples before removing numbers
print("Before Removing Numbers:\n")
print(df["text"].head(5))

# Apply the function
df["text"] = df["text"].apply(remove_numbers)

# Show 5 samples after removing numbers
print("\nAfter Removing Numbers:\n")
print(df["text"].head(5))
```

```python
# Function to remove emojis and special characters
def remove_emojis_special(text):
    return text.encode("ascii", "ignore").decode("ascii")

# Show 5 samples before cleaning
print("Before Removing Emojis & Special Characters:\n")
print(df["text"].head(5))

# Apply the function
df["text"] = df["text"].apply(remove_emojis_special)

# Show 5 samples after cleaning
print("\nAfter Removing Emojis & Special Characters:\n")
print(df["text"].head(5))
```

```python
import nltk
from nltk.corpus import stopwords

# Download NLTK resources
nltk.download('punkt')
nltk.download('stopwords')

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Function to remove stopwords
def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)

# Show 5 samples before removing stopwords
print("Before Removing Stopwords:\n")
print(df["text"].head(5))

# Apply the function
df["text"] = df["text"].apply(remove_stopwords)

# Show 5 samples after removing stopwords
print("\nAfter Removing Stopwords:\n")
print(df["text"].head(5))
```

```python
import re
import string
from nltk.corpus import stopwords

# Load stopwords
stop_words = set(stopwords.words("english"))

# Complete cleaning function
def clean_text(text):
    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove emojis & special characters
    text = text.encode("ascii", "ignore").decode("ascii")

    # Remove stopwords
    words = text.split()
    words = [word for word in words if word not in stop_words]

    return " ".join(words)

# Apply complete cleaning pipeline
df["cleaned_text"] = df["text"].apply(clean_text)

# Display original and cleaned text
print(df[["text", "cleaned_text"]].head(10))
```

```python
import matplotlib.pyplot as plt

# Calculate text length (number of words)
df["text_length"] = df["cleaned_text"].apply(lambda x: len(x.split()))

# Plot histogram
plt.figure(figsize=(8,5))
plt.hist(df["text_length"], bins=20)
plt.title("Histogram of Cleaned Text Lengths")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")
plt.show()

# Print statistics
print("Average Text Length:", df["text_length"].mean())
print("Minimum Text Length:", df["text_length"].min())
print("Maximum Text Length:", df["text_length"].max())

# Observations
print("\nObservations:")
print("1. Most cleaned texts have short to medium length.")
print("2. Stopword removal reduced the number of words.")
print("3. The cleaned dataset is more suitable for NLP models.")
```

```python
import pandas as pd
import re
import string
import nltk
from nltk.corpus import stopwords
from sklearn.preprocessing import LabelEncoder

# Download NLTK resources
nltk.download('punkt')
nltk.download('stopwords')

# Load dataset
df = pd.read_csv("train.txt", sep=";", names=["text", "emotions"])

# Encode emotion labels
le = LabelEncoder()
df["emotion_encoded"] = le.fit_transform(df["emotions"])

# Load stopwords
stop_words = set(stopwords.words("english"))

# Complete text cleaning function
def clean_text(text):
    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove emojis & special characters
    text = text.encode("ascii", "ignore").decode("ascii")

    # Remove stopwords
    words = text.split()
    words = [word for word in words if word not in stop_words]

    return " ".join(words)

# Apply cleaning
df["cleaned_text"] = df["text"].apply(clean_text)

# Save cleaned dataset
df.to_csv("cleaned_emotions.csv", index=False)

# Show value counts of each emotion
print("Emotion Value Counts:")
print(df["emotions"].value_counts())

print("\nCleaned dataset saved successfully as 'cleaned_emotions.csv'")
```
