```python
import pandas as pd
from sklearn.datasets import make_moons

# Create dataset
X, y = make_moons(n_samples=500, noise=0.05, random_state=42)

# Convert to DataFrame
df = pd.DataFrame(X, columns=["Feature1", "Feature2"])

# Display first 10 rows
print("First 10 Rows:")
print(df.head(10))

# Display shape
print("\nDataset Shape:")
print(df.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Apply Standard Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# Show first 5 rows of scaled data
print("Scaled Data (First 5 Rows):")
print(X_scaled[:5])
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Apply KMeans
kmeans = KMeans(n_clusters=2, random_state=42)
df["kmeans_cluster"] = kmeans.fit_predict(X_scaled)

# Scatter Plot
plt.figure(figsize=(6,5))
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=df["kmeans_cluster"], cmap="viridis")
plt.title("KMeans Clustering on Moon Dataset")
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.show()
```

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# Apply DBSCAN
dbscan = DBSCAN(eps=0.3, min_samples=5)
df["dbscan_cluster"] = dbscan.fit_predict(X_scaled)

# Scatter Plot
plt.figure(figsize=(6,5))
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=df["dbscan_cluster"], cmap="plasma")
plt.title("DBSCAN Clustering on Moon Dataset")
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

# KMeans Result
plt.subplot(1,2,1)
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=df["kmeans_cluster"], cmap="viridis")
plt.title("KMeans Clustering")

# DBSCAN Result
plt.subplot(1,2,2)
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=df["dbscan_cluster"], cmap="plasma")
plt.title("DBSCAN Clustering")

plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# Apply DBSCAN with different parameters
dbscan_tuned = DBSCAN(eps=0.2, min_samples=10)
clusters = dbscan_tuned.fit_predict(X_scaled)

# Plot Results
plt.figure(figsize=(6,5))
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=clusters, cmap="rainbow")
plt.title("DBSCAN (eps=0.2, min_samples=10)")
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.show()

# Print unique cluster labels
print("Cluster Labels:", set(clusters))
```

```python
from sklearn.datasets import make_classification
import pandas as pd

# Create High-Dimensional Dataset
X_hd, y_hd = make_classification(
    n_samples=500,
    n_features=10,
    n_informative=8,
    n_redundant=2,
    random_state=42
)

# Convert to DataFrame
df_hd = pd.DataFrame(X_hd, columns=[f"Feature{i+1}" for i in range(10)])

# Display first 10 rows
print(df_hd.head(10))

# Display dataset shape
print("\nDataset Shape:", df_hd.shape)
```

```python
from sklearn.decomposition import PCA
import pandas as pd

# Apply PCA to reduce dimensions from 10 to 2
pca = PCA(n_components=2)
X_pca = pca.fit_transform(df_hd)

# Create DataFrame for PCA result
pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])

# Display first 10 rows
print(pca_df.head(10))

# Print explained variance ratio
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

```python
import matplotlib.pyplot as plt

# Visualize PCA Results
plt.figure(figsize=(6,5))
plt.scatter(pca_df["PC1"], pca_df["PC2"], c=y_hd, cmap="viridis")
plt.title("PCA Visualization")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(label="Class")
plt.show()
```

```python
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Step 1: Create Dataset
X, y = make_moons(n_samples=500, noise=0.05, random_state=42)

# Step 2: Scale Data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 3: Apply DBSCAN
dbscan = DBSCAN(eps=0.3, min_samples=5)
clusters = dbscan.fit_predict(X_scaled)

# Step 4: Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Step 5: Visualize Results
plt.figure(figsize=(6,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=clusters, cmap="rainbow")
plt.title("Complete Pipeline: DBSCAN + PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()

# Step 6: Print Cluster Labels
print("Unique Cluster Labels:", set(clusters))
```
